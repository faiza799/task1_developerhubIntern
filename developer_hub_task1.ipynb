```python
# Import pandas for data handling and analysis
import pandas as pd

# Import seaborn for advanced visualization
import seaborn as sns

# Import matplotlib for plotting graphs
import matplotlib.pyplot as plt
# Load dataset from uploaded file
df = pd.read_csv("iris.csv")
# Check number of rows and columns in dataset
df.shape
# Display column names of dataset
df.columns
# Display first 5 rows to understand data structure
df.head()
# Show dataset information like data types and null values
df.info()
# Get statistical summary (mean, std, min, max, etc.)
df.describe()
# Check for missing values in each column
df.isnull().sum()
# Create scatter plot to show relationship between sepal length and width
# hue='species' colors points based on flower type
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')

# Add title to graph
plt.title("Sepal Length vs Sepal Width Relationship")

# Display the plot
plt.show()
# Plot histogram to show distribution of petal length values
# kde=True adds smooth curve for better understanding
plt.figure(figsize=(6,4))
sns.histplot(df['petal_length'], bins=20, kde=True)

# Add title
plt.title("Distribution of Petal Length")

# Show plot
plt.show()
# Boxplot to check distribution and outliers of petal width
# grouped by species
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x='species', y='petal_width')

# Add title
plt.title("Petal Width Distribution by Species")

# Display plot
plt.show()
```
